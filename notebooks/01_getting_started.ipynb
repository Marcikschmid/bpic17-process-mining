## Einen Fall ansehen
Ein **Case** entspricht einem Prozessdurchlauf. Seine **Events** sind die aufgezeichneten Schritte. Case-Attribute beschreiben den Fall, Event-Attribute einen einzelnen Schritt.

Die folgende Vorschau verwendet den ersten Fall in der Originaldatei. Die Events bleiben in der gespeicherten Reihenfolge. Beachte `lifecycle:transition`: Eine Aktivität kann mehrere Lifecycle-Events haben.

```python
trace = log[0]
pd.Series(dict(trace.attributes), name='Case-Attribute')
```

```python
events = pd.DataFrame([dict(event) for event in trace])
columns = ['concept:name', 'time:timestamp', 'lifecycle:transition', 'org:resource']
events[[column for column in columns if column in events.columns]].head(20)
```
